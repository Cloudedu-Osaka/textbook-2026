## 1. 今日の問い — 「二つの点からの距離の和が一定」 とは何の形か

円は「一つの中心からの距離が一定」 の点の軌跡だった。では **二つの点からの距離の和が一定** にすると、何の形になるか。

> **楕円の定義（焦点による）**: 平面上の二定点 $F_1, F_2$（**焦点 (foci)**）からの距離の和が一定（$|PF_1| + |PF_2| = 2a$、ただし $2a > |F_1F_2|$）である点 $P$ の軌跡を **楕円 (ellipse)** と呼ぶ。

園芸で「二本の杭にひもを掛け、ぴんと張って鉛筆を回す」 と楕円が描けるのは、まさにこの定義である（ひもの長さ $= 2a$ が一定）。この **紐による作図法** は素朴に見えるが、歴史的には「焦点からの距離の和が一定」 という **性質**（Apollonius が前 3 世紀末に証明）を、それを **作図** に翻訳する飛躍であって、明示的な記述はずっと後の **イスラム期（9 世紀、al-Sijzī ら）** を待った。今日われわれが当たり前に使うこの紐は、性質と作図法をつなぐ千年がかりの一手である。今日はこれを作図で確かめ、$F_1, F_2$ を近づけたり離したりすると形がどう変わるか —— 焦点が重なれば円、離れるほど細長く —— を観察する。**二定点・和一定** という最小の約束から、楕円のすべての性質（中心・軸・離心率・準線）が出てくる。

## 3. 定義を作図で確かめる —— 和は一定、二焦点は対称

楕円上に点 $P$ を取り、$|PF_1| + |PF_2|$ が一定であることを確かめる。続いて中心・長軸・短軸の頂点を取り出し、二焦点が中心について対称であることを見る（規律1: 各オブジェクトを一意のシンボルに束縛し、入れ子にしない）。

`````julia
# 和が一定 —— P を楕円の上で動かしても s = 2a = 10 のまま
@ggb P=Point(:el)
@ggb d1=Distance(:P, :F1)
@ggb d2=Distance(:P, :F2)
@ggb s= "d1 + d2"                  # = 2a = 10（P をドラッグして確かめる）
`````

`````julia
# 中心と長軸・短軸の頂点
@ggb M=Midpoint(:F1, :F2)       # 楕円の中心 = 二焦点の中点
@ggb major=Line(:F1, :F2)       # 長軸の直線
@ggb V1=Intersect(:el, :major, 1)   # 長軸の頂点
@ggb V2=Intersect(:el, :major, 2)
@ggb minor=PerpendicularLine(:M, :major)   # 短軸の直線（中心で長軸に直交）
@ggb W1=Intersect(:el, :minor, 1)   # 短軸の頂点
@ggb W2=Intersect(:el, :minor, 2)
`````

`````julia
# 三つの長さ（半径と焦点距離）と関係 a^2 = b^2 + c^2
@ggb a=Distance(:M, :V1)        # 長半径 = 5
@ggb b=Distance(:M, :W1)        # 短半径 = 4
@ggb c=Distance(:M, :F1)        # 焦点距離 = 3
@ggb chk="a^2 - b^2 - c^2"      # = 0 になるはず（a^2 = b^2 + c^2）
`````

`````julia
# 二焦点は中心について対称 —— F1 を中心 M で対称にすると F2 に一致する
@ggb F2chk=Reflect(:F1, :M)
@ggb dsym=Distance(:F2, :F2chk)  # ≈ 0（F1 と F2 は中心について鏡像）
`````

**観察1**: `P` を楕円上で動かしても `s = d1 + d2` は $2a = 10$ から動かない。これが楕円の定義そのもの。

**観察2**: `chk` $\approx 0$、すなわち $a^2 = b^2 + c^2$（$5^2 = 4^2 + 3^2$）。長半径・短半径・焦点距離が直角三角形をなす —— 短軸の頂点 $W_1$ から二焦点までの距離はどちらも $a$（$|W_1F_1| = |W_1F_2| = a = 5$）で、$M$ で直角に切ると $a^2 = b^2 + c^2$ が出る。

**観察3**: `dsym` $\approx 0$。$F_1$ を中心 $M$ で対称にすると $F_2$ に重なる —— **二焦点は中心について互いに鏡像**。前回の「二中心の対称」 が、ここで「二焦点の対称」 になっている。

**問い**: $F_1, F_2$ を近づける（$c \to 0$）と楕円はどうなるか？ 離す（$c \to a$）とどうなるか？ —— 焦点を動かして、形の変化を観察せよ。これが次の §4 の **離心率** につながる。

## 4. 観測量に固める —— 離心率、そして準線への予告

**主命題**: 楕円の長半径 $a$・短半径 $b$・焦点距離 $c$ は $a^2 = b^2 + c^2$ をみたす。二焦点 $F_1, F_2$ は中心 $M$ について対称。楕円の **離心率** を $e = c/a$（$0 \le e < 1$）で定める。

`````julia
# 離心率 e = c/a。焦点を動かすと e が変わり、形が変わる。
@ggb e="c/a"            # = 3/5 = 0.6
`````

```{important} 定義
二焦点 $F_1, F_2$ からの距離の和が一定 $2a$ の点の軌跡を **楕円** と呼ぶ。中心 $M$ は二焦点の中点、$|F_1F_2| = 2c$ を **焦点間距離**、$a$ を **長半径**、$b = \sqrt{a^2 - c^2}$ を **短半径** と呼ぶ。$e = c/a$ を **離心率** と呼び、$e=0$ で円、$e \to 1$ で限りなく細長い楕円になる。
```

```{attention} 二焦点の対称・離心率・準線 —— 前半の伏線がここで一つに結ばれ始める
:class: attention

今日出てきた三つの量は、すべて前半（L6–L9）で仕込んだ伏線の回収である。
```

`````julia
# 焦点・中心・頂点の「値（座標）」 列を見る
cols = [pyconvert(String, x) for x in df.columns]
println("列: ", cols)
df.select(pl.col("Name"), pl.col("Value")).filter(
    pl.col("Name").is_in(pylist(["F1", "F2", "M", "V1", "V2", "W1", "W2"])))
`````

`````julia
# 二焦点の座標を取り出し、中点が中心 M と一致する（= 対称）ことをデータで確認
function coord(df, name)
    row = df.filter(pl.col("Name").is_in([name]))
    val = pyconvert(String, row["Value"][0])           # 点の値文字列（例 "(-3, 0)"）
    nums = [parse(Float64, mm.match) for mm in eachmatch(r"-?\d+\.?\d*", val)]
    return (nums[end-1], nums[end])
end
`````

`````julia
f1 = coord(df, "F1");  f2 = coord(df, "F2");  m = coord(df, "M")
f1, f2, m
`````

`````julia
mid = ((f1[1] + f2[1]) / 2, (f1[2] + f2[2]) / 2)
println("F1 = ", f1, " / F2 = ", f2)
println("二焦点の中点 = ", mid, " / 作図上の中心 M = ", m)
println("一致（= 二焦点は中心について対称）か？ → ", isapprox(mid[1], m[1]; atol=1e-9) && isapprox(mid[2], m[2]; atol=1e-9))
`````

**読み方**: アプレットの内部状態（construction protocol）を DataFrame に取り出すと、$F_1, F_2$ の座標から計算した中点が、作図上の中心 $M$ にぴたり一致する —— **二焦点の対称が、データの上でも閉じる**。L9 では「構造（依存グラフ）の同型」 で対称を見た。今日は「観測量（座標）」 で同じ対称を見た。**構造とデータ、二つの言葉で同じ対称を語れる** ことが、ここまでの道具立ての到達点である。

## 6. 停滞と画期の照射 —— 焦点という一点の、二千年の重層

楕円・放物線・双曲線という名は、紀元前 3 世紀の **Apollonius**（『円錐曲線論』）に遡る。彼は円錐を平面で切るだけでこの三曲線を統一的に記述した —— 今日われわれが「二焦点・和一定」 から入ったのとは別の入り口、**円錐の切断** からである（L11 でこちらの顔に出会う）。

だが「焦点」 という一点は、誰か一人が一度に作ったものではない。**層をなして** 二千年かけて重なってきた。

| 層 | 担い手 | 焦点に与えたもの |
|---|---|---|
| 第 I 層 | **Apollonius**（前 3 世紀末） | 楕円・双曲線で「二焦点からの距離の和（差）が一定」 という **性質を証明**（作図法ではなく、性質として） |
| 第 II 層 | **Pappus**（4 世紀） | 焦点と準線までの距離の比が一定（＝ **離心率**）という焦点・準線性質を初めて記述 |
| 第 III 層 | **イスラム期**（9 世紀、al-Sijzī ら） | 紐による楕円の **作図法**（園丁の作図）を明示的に記述 —— 性質から作図法への飛躍（§1） |
| 第 IV 層 | **Kepler**（1609） | *focus*（ラテン語で「炉・かまど」）という **物理的命名** と、火星軌道で太陽がそこに座るという **天文学的同定** |

ここで一つ、よくある短絡を正しておく。「Kepler が焦点を発明した」 と言われがちだが、これは誤りである。Kepler が「太陽は楕円の中心ではなく、別の特別な点にある」 と気づけたのは、楕円に「中心とは別の特別な点」 があることを **Apollonius からすでに知っていた**（Conics III を参照していた）からである。Kepler の貢献は **点の発見ではなく、物理的命名と天文学的同定** —— 天が円でなく楕円で動き、その焦点に太陽が座るという、円中心の宇宙観を覆す画期だった。

```{admonition} Kepler はどうやって、ほとんど円の火星軌道を楕円と見抜いたか —— 三つの問い
:class: note
Kepler が太陽を焦点に置けたのは（第 IV 層）、火星軌道を **楕円** と同定できたからである。だがこれは、考えるほど不思議な達成である。三つの問いで照らす。

**問い 1 —— 火星軌道は「ほとんど円」である。** 惑星の離心率を比べると:

| 惑星 | 離心率 $e$ | 短半径/長半径 $b/a$ | 太陽の中心からのずれ $c/a$ |
|---|---|---|---|
| 金星 | 0.007 | 0.99998（ほぼ真円） | 0.7% |
| 地球 | 0.017 | 0.99986 | 1.7% |
| **火星** | **0.093** | **0.99563** | **9.3%** |
| 水星 | 0.206 | 0.97864 | 20.6% |

火星は地球の約 5 倍の離心率を持つが、**楕円の「つぶれ」 は長半径に対してわずか 0.44%** —— 目では円と区別できない。金星はさらに円に近く、しかも内惑星で太陽の近くにしか現れず（逆行はするが「衝」 にならない）、Tycho の観測の主軸から外れていた。水星は最も離心率が大きいが太陽に貼りつき観測困難。火星だけが「観測しやすく、かつ離心率の効果が大きい」。だがその効果は、**形（つぶれ 0.44%）ではなく、太陽のずれ（9.3%）と速度変化（最大中心差 ≈ 10.7°）** に現れる。

**問い 2 —— Kepler は彗星を楕円とは思わなかった。** Kepler は 1607 年の彗星（後の **Halley 彗星**）を観測したが、彗星は **直線運動** すると考えていた（『De Cometis』1619）。Halley（1705）との違いは何か。Halley は (a) Newton の万有引力（逆二乗 → 束縛された天体はすべて楕円）を持ち、(b) 1531・1607・1682 年の彗星を **同一天体の約 76 年周期の楕円** と結びつけ、回帰を予言した。Kepler にはこの力学理論も「彗星は戻ってくる」 という発想もなかった。加えて、離心率 $e \approx 0.97$ の彗星は **近日点付近の短い弧** しか観測できず、その弧は **ほとんど直線** に見える —— 一度の通過から軌道の「閉じ」 は読めない。

**問い 3 —— ではなぜ、ほとんど円の火星は楕円と同定でき、極端に細長い彗星は直線とされたのか。** 鍵は「**形を見て決めたのではない**」 ことにある。Kepler は次の四つで楕円に至った。

1. **閉じた全周を、繰り返し観測した** —— Tycho の約 20 年・火星のオポジション約 10 回ぶんのデータは、火星軌道の **全体を何周も** 捉えていた（彗星の一回・短弧とは対照的）。
2. **角度でなく距離を作った** —— 火星が 1 公転（687 日）ごとに軌道上の同じ点へ戻るとき、地球は別の場所にいる。この **地球軌道を動く基線** にして火星の **実距離を三角測量** した。観測された角度を、形を決める量（距離）に変換した。
3. **面積律（速度）が先、形は後** —— Kepler はまず速度変化から **面積律（1602）** を手にし（運動学の中身）、それと整合する **幾何学的形状** を後から探して楕円（1605）に到達した。形を観測から直接帰納したのではなく、**運動学的制約から幾何を逆算** した。
4. **8 分角を捨てなかった** —— 最良の「円 ＋ 等化点（equant）」 模型でも経度に約 **8 分角** の残差が残った。Tycho の精度（1〜2 分角）の数倍であり、Kepler はこれを誤差として切り捨てず、円を手放す根拠とした（後に彼は「この 8 分が天文学の改革を導いた」 と書く）。

→ 楕円は **形として見えた** のではなく、**距離・速度・残差という観測可能な制約から導かれた**。これは本日の主題そのもの —— 楕円を「二焦点・和一定」 という **性質（制約）** から捉えること —— の、天文学における先例である。「ほとんど円」 を楕円と見抜いた Kepler の手つきは、皆さんが今日 `Distance` で和一定を確かめた手つきと、同じ向きを向いている。
```

そして見落とされがちなのが **Pappus**。Apollonius（前 3 世紀）と Kepler（17 世紀）の間の長い時間を埋め、今日の主役である **離心率**（§4）を最初に articulate したのが彼である。今日の $e = c/a$ は、Pappus の第 II 層に直接つながっている。

各層は別々にあるのではなく、互いに支え合っている —— Kepler が太陽を置いた点が、Apollonius が距離和一定で定義した点と「同じ」 であることを保証するには、後の **Dandelin の橋渡し**（L12–L13）を待たねばならなかった。今日の「二焦点・和一定」 は、この重層の **第 I 層** に立っている。

転機の **代数化**（Apollonius が幾何で 1800 年保ったものを、**Descartes / Fermat**（1637 前後）が **座標と方程式** で書き直す解析幾何の誕生）については、次回 L11 で SymPy とともに、より深く扱う。

## 作図で確かめる —— 「ほとんど円」を自分の目で（問い 1 の検証）

§6 の問い 1 を、**実際の離心率で作図** して確かめる。太陽を **共有する焦点** に置き、地球（$e=0.017$）と火星（$e=0.093$）の軌道を本当の比率で描く。YouTube の解説スライドが「円を描いて『だ円』 と書く」 理由を、自分の目で確かめよう。

`````julia
# 太陽を共有焦点に、地球と火星の軌道を実際の離心率で描く。
# 長半径は AU（地球=1）、近日点を +x 方向に揃えた模式（実際の近日点方向は軌道ごとに違う）。
@ggb :const :new
@ggb S=(0, 0)                       # 太陽 ＝ 共有する焦点
@ggb E2=(-0.0334, 0)                # 地球軌道のもう一方の焦点（2c, c=ae=0.0167）
@ggb earth=Ellipse(:S, :E2, 1)      # 地球軌道（長半径 a=1 AU, e=0.0167）
@ggb M2=(-0.2847, 0)                # 火星軌道のもう一方の焦点（2c, c=ae=0.1423）
@ggb mars=Ellipse(:S, :M2, 1.524)   # 火星軌道（長半径 a=1.524 AU, e=0.0934）
`````

`````julia
# 各軌道の中心（二焦点の中点）と、太陽（焦点）からのずれ。
@ggb Ce=Midpoint(:S, :E2)           # 地球軌道の中心
@ggb Cm=Midpoint(:S, :M2)           # 火星軌道の中心
@ggb segm=Segment(:S, :Cm)          # 太陽 → 火星軌道中心 のずれ（目に見える）
@ggb off_e=Distance(:S, :Ce)        # 地球のずれ c = 0.0167（長半径の 1.7%）
@ggb off_m=Distance(:S, :Cm)        # 火星のずれ c = 0.1423（長半径の 9.3%）
`````

`````julia
# 火星軌道の中心に半径 a=1.524 の「真円」を描く。火星の楕円とほぼ重なる（扁平 0.44%）。
@ggb cmp=Circle(:Cm, 1.524)         # 真円。mars 楕円と見分けがつかないはず
@ggb ecc_m="off_m / 1.524"            # 離心率を作図から復元 = c/a ≈ 0.093
`````

**観察1**: `earth` も `mars` も、目には **円** にしか見えない。火星の楕円のつぶれ（短半径/長半径）はわずか 0.44%、地球は 0.01% —— 真円 `cmp` と `mars` は重なって見分けがつかない。

**観察2**: だが太陽 `S` は、各軌道の中心 `Ce` / `Cm` から **ずれている**（`segm` の長さ）。火星では `off_m` $\approx 0.142$ ＝ 長半径の **9.3%**、地球では `off_e` $\approx 0.017$ ＝ **1.7%**。`ecc_m` $\approx 0.093$ に、火星の離心率が作図から戻ってくる。

**読み方**: 楕円の **形**（円からのずれ 0.44%）は見えない。見えるのは **太陽の位置**（中心からのずれ 9.3%）と、その帰結としての速度変化（§6 問い 3）。だから「軌道が楕円だ」 を **形** で示そうとすると円にしかならず、解説スライドのように「円を描いて『だ円』 と書く」 ことになる。Kepler が掴んだのは形ではなく、**太陽のずれと速度** —— 観測可能な制約の方だった。今日の `Distance` で和一定を確かめた手つきは、その縮図である。

```{admonition} 今回の課題
:class: tip

**必修**
1. 二焦点 $F_1, F_2$ と長半径 $a$ を自分で決めて楕円を作図し（$2a > |F_1F_2|$ に注意）、楕円上の点 $P$ をいくつか動かして $|PF_1| + |PF_2| = 2a$ が一定であることを `Distance` で確かめよ。
```

## 枕（次回への予告）—— 円錐を切ると、楕円が現れる

次回は楕円を「作るもの」 から「**切り出すもの**」 へ視点を変える。円錐（アイスクリームのコーン）を、軸に対して斜めの平面でスパッと切ると、切り口は **楕円** になる。今日「二焦点からの距離の和が一定」 と定義した楕円と、円錐の切り口が **同じ形** だというのは、当たり前ではない。その **枕** として、円錐切断の予告だけ置く。

`````julia
# 円錐を斜めの平面で切る（3D）。切り口が楕円になることを目で見る予告。
@ggb :const :new
@ggb cone=Cone((0,0,0), (0,0,1), 0.6)   # 頂点・軸方向・半開角 0.6 rad の円錐
@ggb v = Line((0,0,0),(0.4, 0, 1))
@ggb pl=PerpendicularPlane((0,0,0.5), :v)      # 軸に対して斜めの平面（法線が傾いている）
@ggb section=IntersectConic(:cone, :pl)   # 円錐と平面の交わり = 切断面（斜めなら楕円）
`````

**読み方**: 軸に **垂直** に切れば円、**斜め** に切れば楕円、もっと寝かせれば放物線・双曲線になる。同じ円錐から、平面の傾きだけで三つの曲線が出てくる —— これが Apollonius の統一的視点である。次回は、この切り口の楕円が、今日の「二焦点・和一定」 の楕円と同じものであることを、SymPy で **方程式** に落として確かめる。そして L12 で、その二焦点が **円錐に内接する二つの球（Dandelin 球）の接点** として、円錐の中から必然的に現れる。
