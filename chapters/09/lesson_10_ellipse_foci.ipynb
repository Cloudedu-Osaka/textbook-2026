## 1. 今日の問い — 「二つの点からの距離の和が一定」 とは何の形か

円は「一つの中心からの距離が一定」 の点の軌跡だった。では **二つの点からの距離の和が一定** にすると、何の形になるか。

> **楕円の定義（焦点による）**: 平面上の二定点 $F_1, F_2$（**焦点 (foci)**）からの距離の和が一定（$|PF_1| + |PF_2| = 2a$、ただし $2a > |F_1F_2|$）である点 $P$ の軌跡を **楕円 (ellipse)** と呼ぶ。

園芸で「二本の杭にひもを掛け、ぴんと張って鉛筆を回す」 と楕円が描けるのは、まさにこの定義である（ひもの長さ $= 2a$ が一定）。今日はこれを作図で確かめ、$F_1, F_2$ を近づけたり離したりすると形がどう変わるか —— 焦点が重なれば円、離れるほど細長く —— を観察する。**二定点・和一定** という最小の約束から、楕円のすべての性質（中心・軸・離心率・準線）が出てくる。

## 3. 定義を作図で確かめる —— 和は一定、二焦点は対称

楕円上に点 $P$ を取り、$|PF_1| + |PF_2|$ が一定であることを確かめる。続いて中心・長軸・短軸の頂点を取り出し、二焦点が中心について対称であることを見る（規律1: 各オブジェクトを一意のシンボルに束縛し、入れ子にしない）。

`````julia
# 和が一定 —— P を楕円の上で動かしても s = 2a = 10 のまま
@ggb P=Point(:el)
@ggb d1=Distance(:P, :F1)
@ggb d2=Distance(:P, :F2)
@ggb s=d1 + d2                  # = 2a = 10（P をドラッグして確かめる）
`````

`````julia
# 中心と長軸・短軸の頂点
@ggb M=Midpoint(:F1, :F2)       # 楕円の中心 = 二焦点の中点
@ggb major=Line(:F1, :F2)       # 長軸の直線
@ggb V1=Intersect(:el, :major, 1)   # 長軸の頂点
@ggb V2=Intersect(:el, :major, 2)
@ggb minor=PerpendicularLine(:M, :major)   # 短軸の直線（中心で長軸に直交）
@ggb W1=Intersect(:el, :minor, 1)   # 短軸の頂点
@ggb W2=Intersect(:el, :minor, 2)
`````

`````julia
# 三つの長さ（半径と焦点距離）と関係 a^2 = b^2 + c^2
@ggb a=Distance(:M, :V1)        # 長半径 = 5
@ggb b=Distance(:M, :W1)        # 短半径 = 4
@ggb c=Distance(:M, :F1)        # 焦点距離 = 3
@ggb chk=a^2 - b^2 - c^2        # = 0 になるはず（a^2 = b^2 + c^2）
`````

`````julia
# 二焦点は中心について対称 —— F1 を中心 M で対称にすると F2 に一致する
@ggb F2chk=Reflect(:F1, :M)
@ggb dsym=Distance(:F2, :F2chk)  # ≈ 0（F1 と F2 は中心について鏡像）
`````

**観察1**: `P` を楕円上で動かしても `s = d1 + d2` は $2a = 10$ から動かない。これが楕円の定義そのもの。

**観察2**: `chk` $\approx 0$、すなわち $a^2 = b^2 + c^2$（$5^2 = 4^2 + 3^2$）。長半径・短半径・焦点距離が直角三角形をなす —— 短軸の頂点 $W_1$ から二焦点までの距離はどちらも $a$（$|W_1F_1| = |W_1F_2| = a = 5$）で、$M$ で直角に切ると $a^2 = b^2 + c^2$ が出る。

**観察3**: `dsym` $\approx 0$。$F_1$ を中心 $M$ で対称にすると $F_2$ に重なる —— **二焦点は中心について互いに鏡像**。前回の「二中心の対称」 が、ここで「二焦点の対称」 になっている。

**問い**: $F_1, F_2$ を近づける（$c \to 0$）と楕円はどうなるか？ 離す（$c \to a$）とどうなるか？ —— 焦点を動かして、形の変化を観察せよ。これが次の §4 の **離心率** につながる。

## 4. 観測量に固める —— 離心率、そして準線への予告

**主命題**: 楕円の長半径 $a$・短半径 $b$・焦点距離 $c$ は $a^2 = b^2 + c^2$ をみたす。二焦点 $F_1, F_2$ は中心 $M$ について対称。楕円の **離心率** を $e = c/a$（$0 \le e < 1$）で定める。

`````julia
# 離心率 e = c/a。焦点を動かすと e が変わり、形が変わる。
@ggb e=c/a            # = 3/5 = 0.6
`````

```{important} 定義
二焦点 $F_1, F_2$ からの距離の和が一定 $2a$ の点の軌跡を **楕円** と呼ぶ。中心 $M$ は二焦点の中点、$|F_1F_2| = 2c$ を **焦点間距離**、$a$ を **長半径**、$b = \sqrt{a^2 - c^2}$ を **短半径** と呼ぶ。$e = c/a$ を **離心率** と呼び、$e=0$ で円、$e \to 1$ で限りなく細長い楕円になる。
```

```{attention} 二焦点の対称・離心率・準線 —— 前半の伏線がここで一つに結ばれ始める
:class: attention

今日出てきた三つの量は、すべて前半（L6–L9）で仕込んだ伏線の回収である。

| 今日の量 | 前半の伏線 | 結ばれ方 |
|---|---|---|
| **二焦点の対称**（$F_1 \leftrightarrow F_2$） | L8–L9 内心と傍心の対称（二中心の鏡像） | 二つの中心の対称 → 二つの焦点の対称（軸・中心について鏡像） |
| **離心率** $e = c/a$ | L6 「等距離」 から「距離の比」 へ | 焦点からの距離と準線までの距離の **比** が一定 $= e$（焦点・準線の定義、L11+） |
| **準線（予告）** | L7 「隠れた直線」（Euler 線） | 楕円には焦点ごとに、軸に直交する **隠れた直線（準線 directrix）** があり、$\dfrac{|PF|}{\text{($P$ から準線までの距離)}} = e$ |

L6 で「内接円は三辺から **等距離**」 を見た。楕円はそれを一段ゆるめ、「焦点からの距離と準線までの距離の **比** が一定」 になっている。その比こそ離心率 $e$。**等距離（$e$ が暗に $0$ や $1$）から距離の比（任意の $e$）へ** —— これが円から円錐曲線への拡張である。準線は今日は作図しない（L11 で円錐切断と一緒に立ち上げる）。今日は「焦点・離心率」 と「二焦点の対称」 を手応えとして持てばよい。
```

## 5. ggblab_extra 復習 —— DataFrame で「二焦点の対称」 をデータで確かめる

新しい道具は足さない。L5 で導入した **Construction Protocol → DataFrame** を復習し、作図の内部状態をデータに取り出して、**二焦点が中心について対称** であることを **座標の数値** で確かめる。前回（L9）の自己採点（構造の同型）に対し、今日は **観測量（座標）の対称** をデータで読む —— 構造とデータの両面から「対称」 を捉える練習である。

`````julia
# L5/L7/L8 と同じ作法（Julia から Python の道具を借りる）
using PythonCall
pl = pyimport("polars")
ggblab_extra = pyimport("ggblab_extra")
ConstructionIO = ggblab_extra.ConstructionIO
`````

`````julia
# 作図全体を DataFrame に取り出す（L5 §… と同じ手順）
df = @await ConstructionIO.initialize_dataframe(ggb, use_applet=true)
# 焦点・中心・頂点の「値（座標）」 列を見る
cols = [pyconvert(String, x) for x in df.columns]
println("列: ", cols)
df.select(pl.col("Name"), pl.col("Value")).filter(
    pl.col("Name").is_in(pylist(["F1", "F2", "M", "V1", "V2", "W1", "W2"])))
`````

`````julia
# 二焦点の座標を取り出し、中点が中心 M と一致する（= 対称）ことをデータで確認
function coord(df, name)
    row = df.filter(pl.col("Name") == name)
    val = pyconvert(String, row["Value"][0])           # 点の値文字列（例 "(-3, 0)"）
    nums = [parse(Float64, mm.match) for mm in eachmatch(r"-?\d+\.?\d*", val)]
    return (nums[1], nums[2])
end

f1 = coord(df, "F1");  f2 = coord(df, "F2");  m = coord(df, "M")
mid = ((f1[1] + f2[1]) / 2, (f1[2] + f2[2]) / 2)
println("F1 = ", f1, " / F2 = ", f2)
println("二焦点の中点 = ", mid, " / 作図上の中心 M = ", m)
println("一致（= 二焦点は中心について対称）か？ → ", isapprox(mid[1], m[1]; atol=1e-9) && isapprox(mid[2], m[2]; atol=1e-9))
`````

**読み方**: アプレットの内部状態（construction protocol）を DataFrame に取り出すと、$F_1, F_2$ の座標から計算した中点が、作図上の中心 $M$ にぴたり一致する —— **二焦点の対称が、データの上でも閉じる**。L9 では「構造（依存グラフ）の同型」 で対称を見た。今日は「観測量（座標）」 で同じ対称を見た。**構造とデータ、二つの言葉で同じ対称を語れる** ことが、ここまでの道具立ての到達点である。

## 6. 停滞と画期の照射 —— Apollonius から Kepler へ、そして代数へ

楕円・放物線・双曲線という名は、紀元前 3 世紀の **Apollonius**（『円錐曲線論』）に遡る。彼は円錐を平面で切るだけでこの三曲線を統一的に記述した —— 今日われわれが「二焦点・和一定」 から入ったのとは別の入り口、**円錐の切断** からである（L11 でこちらの顔に出会う）。だが Apollonius の記述は **斜交座標系による局所的なもの** で、「焦点」 という言葉も概念も、まだ前面には出ていなかった（焦点を表す語 *focus*〔ラテン語で「炉・かまど」〕は、ずっと後に Kepler が与えたものである）。

転機は二つ。一つは **Kepler**（1609）が、火星の軌道を円ではなく **楕円** とし、太陽がその **一方の焦点** にあると見抜いたこと —— 天が円でなく楕円で動くという、円中心の宇宙観を覆す画期だった。もう一つは **Descartes / Fermat**（1637 前後）が、Apollonius が幾何で 1800 年保ったものを **代数（座標と方程式）** で書き直したこと —— 「観察された曲線」 を $x^2/a^2 + y^2/b^2 = 1$ という **方程式** に移す、解析幾何の誕生である。

皆さんは今日、楕円を **作図（観察）** から入った。L11 で **円錐の切断** という別の顔に出会い、SymPy で **方程式** へ移す（Descartes の画期の現代的再演）。そして L12–L13 で、二焦点が **Dandelin 球の二つの接点** であること —— 今日「定義」 として天下りに置いた二焦点が、円錐の中に **必然として現れる** こと —— を見る。今日の「二焦点・和一定」 は、その長い回収の出発点である。

```{note} 伏線回収 —— 「二中心の対称」 から「二焦点の対称」、そして Dandelin へ
:class: note
L8–L9 で発見した「内心と傍心が辺について対称」 は、今日「二焦点が中心について対称」 として円錐曲線に持ち上がった。だが今日の二焦点は、まだ「和が一定になるように置いた二点」 という **定義** にすぎない。

L12–L13 で、この二焦点が **円錐を切る平面に内側から接する二つの球（Dandelin 球）の接点** であることが分かる。二つの球は切断面の上下から接し、互いに鏡像 —— だから二焦点も鏡像になる。今日「なぜ二つあるのか・なぜ対称なのか」 を天下りに受け取った二焦点が、円錐の三次元構成から **必然** として立ち上がる。L6（等距離）・L7（隠れた直線＝準線）・L8–L9（内外の対称）・L10（二焦点）の伏線は、すべて Dandelin の二球で一つに結ばれる。焦点・準線・離心率・Dandelin という語は、**今は分からなくて正常**。L11–L13 で必ず戻ってくる。
```

## 7. 次回への接続

今日は楕円を **二焦点・和一定** という素朴な定義から作図し、中心・半径・焦点距離・離心率という観測量を手にした。だがこれは楕円の **一つの顔** にすぎない。

次回（第 11 回）は、楕円の **もう一つの顔** —— **円錐を平面で切ると現れる切断面** —— に出会う。観察された切断面が、なぜ今日の「二焦点・和一定」 の楕円と同じものなのか。その問いに答えるため、観察を **方程式 $x^2/a^2 + y^2/b^2 = 1$** へ移す道具として **SymPy（記号計算）** を初めて導入する（ggblab_extra 第四機能の初出）。Apollonius が幾何で保ったものを Descartes が代数で書き直した画期を、現代のツールで再演する。

## 枕（次回への予告）—— 円錐を切ると、楕円が現れる

次回は楕円を「作るもの」 から「**切り出すもの**」 へ視点を変える。円錐（アイスクリームのコーン）を、軸に対して斜めの平面でスパッと切ると、切り口は **楕円** になる。今日「二焦点からの距離の和が一定」 と定義した楕円と、円錐の切り口が **同じ形** だというのは、当たり前ではない。その **枕** として、円錐切断の予告だけ置く。

`````julia
# 円錐を斜めの平面で切る（3D）。切り口が楕円になることを目で見る予告。
@ggb :const :new
@ggb cone=Cone((0,0,0), (0,0,1), 0.6)   # 頂点・軸方向・半開角 0.6 rad の円錐
@ggb pl=Plane((0,0,2), (0.4, 0, 1))      # 軸に対して斜めの平面（法線が傾いている）
@ggb section=IntersectConic(:cone, :pl)   # 円錐と平面の交わり = 切断面（斜めなら楕円）
`````

**読み方**: 軸に **垂直** に切れば円、**斜め** に切れば楕円、もっと寝かせれば放物線・双曲線になる。同じ円錐から、平面の傾きだけで三つの曲線が出てくる —— これが Apollonius の統一的視点である。次回は、この切り口の楕円が、今日の「二焦点・和一定」 の楕円と同じものであることを、SymPy で **方程式** に落として確かめる。そして L12 で、その二焦点が **円錐に内接する二つの球（Dandelin 球）の接点** として、円錐の中から必然的に現れる。
