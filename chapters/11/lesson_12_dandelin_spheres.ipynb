# 第12回: Dandelin 球 — 二焦点はなぜ二つあり、なぜ「距離の和が一定」なのか（円錐の中の必然）

**主題**: L10 で楕円を「**二焦点からの距離の和が一定**」と **定義** し、L11 で円錐の切り口が同じ方程式に落ちることを SymPy で確かめた。だが L11 の最後に正直に残した問いがあった —— **なぜ切り口の二焦点が、定義の二焦点と同じ場所に来るのか**。今日はその「なぜ」に、**Dandelin 球**（円錐に内側から接する二つの球）で答える。切断面に上下から接する二つの球の **接点** が、まさに二焦点である。「二焦点が二つあること」も「距離の和が一定であること」も、円錐の三次元構成から **必然** として立ち上がる —— 天下りだった定義が、立体の幾何に接地する回である。

**副題**: 今日の道具立ては「**立体で見る → 接点として取り出す → 長さの等しさで言い切る**」。新しいソフトウェア機能は足さない（L11 までの ggblab・SymPy で足りる）。足すのは **一つの古典的アイデア** —— *球外の一点から球へ引いた二本の接線の長さは等しい*（L4 の円の接線、円の冪の三次元版）。この一手だけで、二千年あいだ別々だった「円錐の切り口」と「二焦点・和一定」が **一つ** に結ばれる。1822 年、Germinal Dandelin の作図的一撃である（§6）。

## 0. 前回 (第 11 回) の振り返り — 「同じ方程式」までは来たが、「なぜ」は残った

前回は、円錐を斜めに切った切り口の楕円が、L10 で「二焦点・和一定」と定義した楕円と **同じ方程式 $x^2/a^2 + y^2/b^2 = 1$** に落ちることを SymPy で確かめた。焦点 $(\pm c, 0)$・離心率 $e=c/a$・準線 $x=\pm a^2/c$ が、方程式の側から計算で戻ってきた。Apollonius の幾何を Descartes の代数で書き直す画期（解析幾何）の再演だった。

## 1. 今日の問い — 切り口の二焦点は、どこから来るのか

L11 までで、楕円には二つの顔があった —— 「二焦点からの距離の和が一定」（定義）と「円錐の斜めの切り口」（観察）。同じ方程式に落ちることは確かめた。だが切り口を見ても、**焦点はどこにも描かれていない**。切断面はただの楕円の縁であって、その内側の二点（焦点）が立体のどこから来るのかは、見えないままだった。

> **今日の問い**: 円錐の切り口（楕円）の **二焦点** は、円錐の三次元構成の **どこ** にあるのか。そして「距離の和が一定」は、立体の幾何から **必然** として出てくるのか。

答えは、円錐の中に **二つの球** を入れることにある。切断面に上下から接するように球を膨らませると、ちょうど一点で切断面に触れる球が、上下に一つずつ取れる。その **二つの接点** が二焦点である —— というのが Dandelin（1822）の発見だった。今日はそれを作図で再現し、なぜそうなるのかを「接線の長さの等しさ」一つで説明する。

## 3. 円錐に二つの球を入れる —— 接点が焦点になる

### 3.1 切断面に上下から接する二つの球

円錐の内壁にぴったり接しながら（＝円錐に内接しながら）膨らむ球を考える。中心を軸の上に置くと、球は円錐の内壁に **一つの円** で接する。その球を切断面に **触れる**まで膨らませると、切断面の **下側** に一つ、**上側** に一つ、ちょうど接する球が取れる。これが **Dandelin 球** である。

軸上に中心を置いた半径 $r$ の球が円錐に内接する条件は、中心の高さ $h$（頂点からの距離）に対して $r = h\sin\alpha$（$\alpha$ は半開角）。さらに切断面に接する条件（中心から切断面までの距離 $=r$）を課すと、上下それぞれの球の中心と半径が **一意に決まる**。下の式の値は、その二条件を解いたものである（導出は思考課題3）。

`````julia
# 切断面に上下から接する二つの Dandelin 球。中心は軸上、半径は「円錐に内接 かつ 切断面に接する」
# 二条件から確定した値（思考課題3で導出）。中心を名前付き点にしておく（規律1）。
@ggb C1=Point("{0, 0, 1.8257}")            # 下の球の中心（軸上、切断面より下）※点は Point("{…}") で
@ggb sph1=Sphere(:C1, 0.6260)              # 下の Dandelin 球（切断面に下から接する）
@ggb C2=Point("{0, 0, 3.9639}")            # 上の球の中心（軸上、切断面より上）
@ggb sph2=Sphere(:C2, 1.3592)              # 上の Dandelin 球（切断面に上から接する）
`````

**観察2**: 二つの球は **大きさが違う**（下 $r_1 \approx 0.63$、上 $r_2 \approx 1.36$）。これは斜めの切り口では避けられない —— 頂点に近い下の球は細い場所に入るので小さく、上の球は広い場所なので大きい。**二つの球は「鏡像」ではない**。軸に垂直に切れば（円のとき）二球は同じ大きさになるが、斜めにした瞬間に非対称になる。この点はあとで「二焦点の対称」を語るとき（§5）に効いてくる。

### 3.2 接点を焦点として取り出す —— 中心から切断面への垂線の足

球が平面に接する点は、**球の中心から平面へ下ろした垂線の足** である（接点では半径が平面に直交するから）。これを作図でそのまま取る。

`````julia
# 各球の中心から切断面へ垂線を下ろし、その足（= 接点）を取る。これが焦点になる。
@ggb perp1=PerpendicularLine(:C1, :pl)     # C1 から切断面 pl への垂線
@ggb F1=Intersect(:perp1, :pl)             # 垂線の足 = 下の球と切断面の接点 = 焦点その1
@ggb perp2=PerpendicularLine(:C2, :pl)     # C2 から切断面への垂線
@ggb F2=Intersect(:perp2, :pl)             # 上の球と切断面の接点 = 焦点その2
`````

`````julia
# F1, F2 が本当に切り口の楕円 section の焦点か —— 中心 M（長軸の中点）と並べて見る。
@ggb M=Midpoint(:F1, :F2)                  # 二接点の中点（= 楕円の中心になるはず）
`````

**観察3**: `F1`, `F2` は切断面の上、楕円 `section` の内側に落ちる。座標はおよそ $F_1 \approx (0.23, 0, 2.41)$, $F_2 \approx (-0.50, 0, 2.70)$、中点 $M \approx (-0.14, 0, 2.55)$。この二点は、**この切り口の楕円の二焦点** —— すなわち L10 の定義「距離の和が一定」を満たす二点（次の §3.3 で確認）であり、L11 §3 のように方程式に落として計算しても同じ二点が出る（L11 §3 の数値例は $a=5$ の別の楕円なので座標値は違うが、手順は同じ）。**画面のどこにも描かれていなかった焦点が、二つの球の接点として立体から現れた**。

### 3.3 距離の和は一定 —— $P$ をどこに動かしても $2a$

楕円の定義（L10）は「$|PF_1| + |PF_2| = 2a$」だった。切り口の楕円 `section` の上に点 $P$ を取り、いま立体から取り出した二接点 $F_1, F_2$ について、この和が一定であることを確かめる。

`````julia
# 切り口の楕円上の点 P をドラッグしても、二接点までの距離の和が一定であることを見る。
@ggb P=Point(:section)                     # 切り口の楕円の上の点（ドラッグできる）
@ggb d1=Distance(:P, :F1)
@ggb d2=Distance(:P, :F2)
@ggb s="d1 + d2"                             # = 2a ≈ 2.009（P をどこに動かしても一定）
`````

**観察4**: `P` を楕円 `section` の上でどこに動かしても、`s = d1 + d2` は $2a \approx 2.009$ から動かない。**L10 で「紐の長さ一定」として天下りに置いた和一定が、ここでは「二つの球の接点までの距離の和」として、円錐の中から出てきた**。だが「なぜ一定なのか」はまだ作図が示しただけ。次の §3.4 でその理由を言い切る。

### 3.4 なぜ和が一定か —— 接線の長さの等しさ + 母線に沿った一定距離

理由は、**たった一つの古典的事実**から出る。

> **球の接線の長さ（L4 の円の接線の三次元版）**: 球の外の一点 $P$ から球に引いた接線（接点までの線分）の長さは、**どの方向に引いても等しい**。

切り口の点 $P$ は円錐の表面にある。$P$ を通る **母線**（頂点から $P$ を通ってまっすぐ伸びる円錐表面の直線）を考える。この母線は、下の球が円錐に接する円 $k_1$ を点 $A$ で、上の球が接する円 $k_2$ を点 $B$ で貫く。すると：

- $|PF_1|$ と $|PA|$ は、どちらも **$P$ から下の球への接線** —— $PF_1$ は切断面での接点 $F_1$ まで、$PA$ は円錐に沿った接点 $A$ まで。同じ球への二接線だから **$|PF_1| = |PA|$**。
- 同様に $|PF_2| = |PB|$（上の球への二接線）。

したがって、

$$|PF_1| + |PF_2| = |PA| + |PB| = |AB|.$$

$A$ は接触円 $k_1$（高さ一定）の上、$B$ は接触円 $k_2$（高さ一定）の上にあり、両方とも **同じ母線** の上にある。母線に沿った $k_1$ と $k_2$ のあいだの長さ $|AB|$ は、**どの母線でも同じ**（円錐の回転対称性）。ゆえに $|PF_1| + |PF_2| = |AB| = $ 一定 $= 2a$。これが「距離の和が一定」の **理由** である。

`````julia
# §3.4 を数値で確かめる —— 二つの接触円のあいだの「母線に沿った長さ |AB|」が、
# どの母線（角度 θ）でも一定であることを Julia で計算する（カーネルは Julia）。
α = 0.35
ta, ca = tan(α), cos(α)
z1c, z2c = 1.6111, 3.4979        # 二つの接触円の高さ（球が円錐に接する円の z）
for θ in (0.0, 1.0, 2.0, 3.5, 5.0)
    A = (z1c*ta*cos(θ), z1c*ta*sin(θ), z1c)   # 下の接触円 k1 上の点（母線 θ）
    B = (z2c*ta*cos(θ), z2c*ta*sin(θ), z2c)   # 上の接触円 k2 上の点（同じ母線 θ）
    AB = sqrt(sum((B .- A).^2))
    println("θ = $θ:  |AB| = ", round(AB, digits=5))
end
`````

**観察5**: `|AB|` は $\theta$（どの母線か）に依らず $\approx 2.00856$ で一定 —— これがそのまま $2a$ である。母線に沿った接点間の長さが一定だから、$P$ をどこに置いても $|PF_1| + |PF_2|$ が同じになる。**和一定は「紐の長さ」という約束ではなく、円錐の回転対称性 ＋ 接線の等長性から出る幾何の必然だった**。L10 で `Distance` を足し算した手つきと、L11 で根号がほどけて $2a$ になった式（$d_1 = a + \tfrac{c}{a}x$ など）が、ここで立体の一本の母線として **意味** を持つ。

## 4. 観測量に固める —— 二焦点 = 二接点、和一定 = 母線長

**主命題（Dandelin）**: 円錐を斜めの平面で切るとき、切断面に内側から接する二つの球（Dandelin 球）が上下に一意に定まる。各球と切断面の **接点 $F_1, F_2$** は、切り口の楕円の **二焦点** である。切り口の点 $P$ について $|PF_1| + |PF_2|$ は、二つの球が円錐に接する円のあいだの **母線に沿った長さ $|AB|$ に等しく、$P$ によらず一定 $=2a$** である。

```{important} 定義・命題のまとめ（L10–L12 を一つに）
- **二焦点**（L10 の定義） ＝ Dandelin 二球と切断面の **二つの接点**（L12 の立体）。
- **距離の和が一定 $2a$**（L10） ＝ 二つの接触円のあいだの **母線長 $|AB|$**（L12）。
- 球が円錐に内接する条件 $r = h\sin\alpha$ と、切断面に接する条件（中心-平面間距離 $=r$）の二つで、上下の球の中心・半径が一意に決まる。
- 斜めの切り口では **上下の球は大きさが違う**（$r_1 \ne r_2$）。軸に垂直（円）のときだけ等しい。
```

```{attention} 立体 → 接点 → 長さ —— 「定義」が「必然」に変わった
:class: attention

L10 で楕円を **定義**（距離の和が一定）から入り、L11 で **方程式**（$x^2/a^2+y^2/b^2=1$）に固定した。今日 L12 で、その定義が **円錐の三次元構成からの必然** に変わった。

| 回 | 楕円をどう捉えたか | 二焦点はどこにあるか |
|---|---|---|
| **L10** | 二焦点からの距離の和が一定（定義・天下り） | 紐の二つの杭（与えるもの） |
| **L11** | $x^2/a^2+y^2/b^2=1$（方程式・SymPy） | $(\pm c, 0)$（式から計算で出る） |
| **L12** | 円錐の切り口 ＝ Dandelin 二球の接点（立体・必然） | 二つの球が切断面に触れる点（立体から現れる） |

「天下りに与えた二点」が「式から計算される二点」になり、ついに「**円錐の中から必然として現れる二点**」になった。同じ二焦点を、約束・式・立体の三つの言葉で語れる —— これが L10–L12 の到達点である。
```

## 5. 二焦点の対称は、どこから来るのか —— 球の合同 *ではない*

L8–L9 で「内心と傍心が辺について対称」を発見し、L10 でそれを「二焦点が中心 $M$ について対称」として持ち上げた。その対称は今日も正しい（$F_1$ を $M$ で対称にすると $F_2$ に重なる）。だが **その理由を取り違えてはいけない**。

二焦点が対称なのは、**二つの球が鏡像だから ではない** —— §3.2 で見たとおり、上下の球は大きさが違う。では何が対称を生むのか。**図全体が、長軸（と円錐の軸）を含む一枚の平面について左右対称**だからである。いまの作図では、円錐の軸（$z$ 軸）と切断面の傾き方向（$x$ 方向）が作る $xz$ 平面が、その対称面になっている。この鏡映で円錐・切断面・上下二球はそれぞれ自分自身に写り、楕円は長軸について折り返される。焦点 $F_1, F_2$ はこの長軸上に乗り、**長軸の中点 $M$ について対称**になる。

もう一段だけ、答えを言わずに **入口** を開けておく。二つの球はどちらも「切断面」と「円錐の母線」の両方に接している。球が二つのものに接するなら、その中心は両者の **なす角を二等分する** 位置に来る（L4 で「円の中心は二接線の角の二等分線上」と見たのと同じ理屈）。上下の球は、母線と切断面のなす角を、切断面を挟んで **一方は内側から・もう一方は外側から** 二等分する位置に、それぞれ座る。**「内側からの二等分」と「外側からの二等分」が対で現れて、辺（切断面）について対称な二点（＝二焦点）を生む** —— この構造を、前半（L8–L9）のどこかで見なかっただろうか。二本の角の二等分線が一点で交わり、内側と外側の二等分が対の中心を作った、あの図である。答えは今日は書かない。今回の思考課題で、二次元（三角形）と三次元（円錐）を自分の手で重ねてほしい。

```{note} 取り違えやすい「対称」—— 名前や見た目でなく構造で帰属する（L9 の作法）
:class: note
「二球が上下に一つずつあるのだから鏡像だろう」 と思い込むと誤る（斜めの切り口では球の大きさが違う）。二焦点の対称は **球の合同** からではなく、**図全体の鏡映対称面**（長軸を含む面）から来る。L9 で「座標や見た目でなく、構造（依存グラフの同型）で型を帰属する」 採点を練習した —— ここでも、見た目の「上下一対」 に引きずられず、どの対称操作が図を不変に保つかで対称を帰属する。これが L8–L11 から続く「構造で見る」 一貫した作法である。
```

**伏線回収（到達点）**: L6（等距離）→ L7（隠れた直線＝準線）→ L8–L9（内外の二中心の対称）→ L10（二焦点・和一定）→ L11（方程式）と仕込んできた糸が、ここで円錐の三次元構成として一つに結ばれた。「二つあること」は二つの球（上下）から、「和が一定」は母線長 $|AB|$ から、「対称」は図の鏡映面から —— それぞれ別の理由で、しかし同じ一つの立体の中から出てくる。

## 6. 停滞と画期の照射 —— 1822 年、総合幾何の最後の宝石

円錐の切り口（Apollonius、前 3 世紀）と「二焦点・距離の和が一定」（同じく Apollonius が性質として証明）は、二千年のあいだ **別々の顔** として知られていた。両者が「同じ楕円」 であることは、L11 で見たように Descartes/Fermat 以降は **代数** で確かめられた —— 座標を立て、両方を方程式に落とし、一致を計算する。だがそれは「式が一致した」 のであって、**幾何的な理由** を見せるものではなかった。

その理由を、座標も方程式も使わず、**純粋に総合幾何（作図的議論）だけ** で一撃で示したのが、**Germinal Pierre Dandelin**（1794–1847、ベルギーの数学者・技師）である（1822）。彼の二つの球は、§3.4 で見たとおり「接線の長さの等しさ」 という古典的事実一つで、切り口と二焦点を結んだ。同時代の **Adolphe Quetelet**（統計学で有名だが、若い頃は幾何学者）も近い結果に到達しており、準線まで含めた形は **Dandelin–Quetelet の定理** と呼ばれる（準線は L13）。

これは数学史の文脈で見ると印象深い位置にある。1822 年は、解析幾何（Descartes）と微積分（Newton/Leibniz）が幾何学の主役になって久しく、**総合幾何はもう古い道具** と見なされ始めていた時期である。そこへ Dandelin は、座標を一切使わない作図的議論で、二千年の二つの顔を結ぶ宝石のような証明を出した —— いわば **総合幾何が最後に見せた切れ味** である。

そして、ここが一番明るい含意である。Dandelin の球は、円錐・平面・球というユークリッド幾何の範囲だけでできていて、**古代ギリシア人が見つけていてもおかしくなかった** —— それでも見つかるまでに **二千百年** かかった。同じことが二十世紀にも起きる。**Feynman**（1964・失われた講義）は、Newton の逆二乗則から Kepler の楕円軌道を、微積分を使わず **初等幾何だけ** で導く「円形ホドグラフ／円形準線」を組み立てた（L13–L15 で扱う）。これも Newton 自身が見つけていてもおかしくない構成で、やはり二世紀以上見過ごされていた。円錐曲線という **二千三百年前に始まった** 分野で、19 世紀（Dandelin）と 20 世紀（Feynman）に、初等幾何のレベルの新発見があった —— **二千年をかけても発見できない美しい真実が、まだ残っている**。数学は完成された遺産ではなく、いまも開いている営みである。次に二千年見落とされてきた構成を見つけるのは、この教室の誰かかもしれない。

## 枕（次回への予告）—— 接触円が、隠れた直線（準線）を呼ぶ

次回は、二つの Dandelin 球が円錐に接する **二つの円（接触円 $k_1, k_2$）** に注目する。その円を含む平面と、切断面との **交わりの直線** —— それが準線である。焦点（接点）と準線（接触円の平面との交わり）が、どちらも Dandelin 球から出てくる。その **枕** として、下の球の接触円だけ予告で描く。

`````julia
# 下の Dandelin 球が円錐に接する円（接触円 k1）の予告。高さ z ≈ 1.611、半径 ≈ 0.588。
# 次回、この円を含む平面と切断面の交わりとして「準線」が立ち上がる。
@ggb nrm0=Vector((0, 0, 1))                # 円の面の法線（軸方向）= 水平な円
@ggb Ck1=Point("{0, 0, 1.6111}")           # 接触円の中心（軸上）※点は Point("{…}") で
@ggb k1=Circle(:Ck1, 0.5881, :nrm0)        # 接触円（中心・半径・面の法線）
`````

| 辺 $BC$ の傍接円 | 切断面 下側の Dandelin 球 |
