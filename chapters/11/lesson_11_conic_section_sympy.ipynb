## 0. 前回 (第 10 回) の振り返り — 楕円を「作った」

前回は三角形（前半・五心）を離れ、楕円に出会った。楕円を「**二定点 $F_1, F_2$（焦点）からの距離の和が一定 $2a$**」という定義から **作図** し、中心 $M$・長半径 $a$・短半径 $b$・焦点距離 $c$ を取り出して $a^2 = b^2 + c^2$ を確かめ、**離心率 $e = c/a$** を観測量として手にした。二焦点が中心について **対称**（L8–L9 の「二中心の対称」の続き）であることも見た。

今日は、その楕円を「**作るもの**」から「**切り出すもの**」へ視点を変える。そして「切り出した楕円」と「作った楕円」が **同じ方程式** であることを、SymPy で確かめる。L10 で手にした観測量（$a, b, c, e$）はそのまま持っていく —— 今日の主役は新しい幾何ではなく、**観察を式に移す道具**である。

## 1. 今日の問い — 円錐の切り口は、本当にあの楕円か

L10 の最後（枕）で、円錐を斜めに切ると切り口が楕円に「見える」ことを予告した。だが **見える** と **同じである** は違う。

> **今日の問い**: 円錐を斜めの平面で切った切り口は、L10 で「二焦点・和一定」と定義した楕円と、本当に **同じもの** か。同じだとして、それを **どうやって確かめる** か。

「目で見て楕円っぽい」では確かめたことにならない。確かめるには、切り口を **方程式** に落とし、その方程式が $x^2/a^2 + y^2/b^2 = 1$ の形をしていて、そこから出てくる焦点・離心率が L10 のものと一致することを **計算で示す** 必要がある。これが今日 SymPy を導入する動機である —— **観察を、検証できる式に固定する**。

これは数学史の一大画期の再演でもある。Apollonius（紀元前 3 世紀）は円錐を切るだけで楕円・放物線・双曲線を **幾何** で統一的に記述し、それが 1800 年保った。だが「焦点」を式から取り出すような操作はできなかった。Descartes / Fermat（1637 前後）が、その幾何を **座標と方程式（代数）** で書き直して初めて、曲線を「計算する」道が開けた。今日はその移行を、現代のツール（SymPy）で手元に再現する。

```{admonition} 補足 —— 円錐曲線はどこから来たか（Apollonius よりさらに古く）
:class: note
円錐曲線は Apollonius が始めたのではない。さらに古く、**Menaechmus**（前 4 世紀）が **立方体倍積問題**（与えられた立方体の体積を 2 倍にする辺を作る）を解くために、$a : x = x : y = y : 2a$ という **二つの比例中項** を $x^2 = ay$ と $y^2 = 2ax$ の交点として視覚化した —— このとき円錐曲線が「比例中項を見る道具」 として歴史に登場する。Menaechmus の二曲線は、問題の対称構造から **直交する軸** を共有していた。

ただし Menaechmus が持っていたのは **比例関係** であって、文字 $x, y$ や等号「$=$」 を使う **方程式** ではない（文字代数は 16 世紀の Viète、「$=$」 は Recorde を待つ）。Apollonius の革新は「**どの直径を選んでもよい**」（アフィン不変性）を見出したことだが、その代償に **問題由来の特権的な直交軸が消え**、曲線は問題から切り離された抽象的対象になった。今日われわれが座標 $x, y$ を立てて方程式に落とすのは、Menaechmus 的な **問題由来の直交軸** を、Descartes 的な代数で取り戻す動きでもある。
```

## 3. ggblab_extra（第四機能・初出）—— SymPy で観察を式に移す

ここで新しい道具を一つだけ足す。L5 で作図を **DataFrame**（データ）に、L7 で **有向グラフ**（構造）に外部化し、L9 でその構造で **自己採点** した。今日はもう一段、作図を **方程式（記号）** に移す —— **SymPy（記号計算ライブラリ）** を導入する。

### 3.1 なぜ今これが必要か

§2 で切り口が「楕円に見える」ことは観察した。だが「見える」を「L10 の楕円と同じだ」に変えるには、座標でも見た目でもなく **方程式** が要る。方程式 $x^2/a^2 + y^2/b^2 = 1$ に落ちれば、そこから焦点 $(\pm c, 0)$ と離心率 $e=c/a$ が **計算で** 出てくる。それが L10 の二焦点・離心率と一致すれば、「同じ楕円だ」と **言い切れる**。SymPy は、この「観察 → 式 → 焦点・離心率」の移行を機械的に・正確に実行する道具である。

### 3.2 何が起きているか（最小デモ）

`````julia
# L5/L7/L9 と同じ作法（Julia から Python の道具を借りる）。今日は SymPy。
using PythonCall
sympy = pyimport("sympy")
geo = pyimport("sympy.geometry")
`````

`````julia
# L10 の楕円（長半径 a=5・短半径 b=4）を SymPy の楕円として持ち、その「方程式」を見る。
el_s = geo.Ellipse(geo.Point(0, 0), 5, 4)   # 中心(0,0)・長半径 5・短半径 4
eq = el_s.equation()                          # → x^2/25 + y^2/16 - 1（= 0 の形）
`````

`````julia
# 方程式から焦点・離心率・焦点距離を SymPy が計算で取り出す。
foci  = el_s.foci             # → (Point2D(-3,0), Point2D(3,0)) = L10 の F1, F2
ecc   = el_s.eccentricity     # → 3/5 = L10 の離心率 e
fdist = el_s.focus_distance   # → 3 = L10 の焦点距離 c
println("焦点 = ", foci)
println("離心率 e = ", ecc, " / 焦点距離 c = ", fdist)
`````

**観察2**: 方程式 $x^2/25 + y^2/16 - 1 = 0$ から、SymPy は焦点 $(\pm 3, 0)$・離心率 $3/5$・焦点距離 $3$ を **計算で** 返す。これは L10 で作図から取り出した $F_1=(-3,0)$, $F_2=(3,0)$, $e=0.6$, $c=3$ と **一致** する。**作図（観察）で得たものが、方程式（式）の側からも同じものとして戻ってきた** —— これが Descartes の画期の手触りである。

### 3.3 定義（距離の和）から方程式が出る —— 根号がほどける瞬間

L10 の定義は「$|PF_1| + |PF_2| = 2a$」だった。これを SymPy で式にすると、**根号がきれいにほどけて** 和が $2a$ に一定化する。楕円上では $y^2 = b^2(1 - x^2/a^2)$（方程式を $y^2$ について解いた形）であることを使う。

`````julia
# 定義の二つの距離を式にする。F1=(-c,0), F2=(c,0)、楕円上なので y^2 = b^2(1 - x^2/a^2)。
# using SymPyPythonCall   # Julia ネイティブに扱える記号。`x^2` などの演算子がそのまま効く。
# @syms x                 # 記号 x を用意（sympy.symbols("x") の Julia 版）
x = sympy.Symbol('x', real=true)
`````

`````julia
a = 5; c = 3
b2  = a^2 - c^2               # = b^2 = 16
ysq = b2 * (1 - x^2 / a^2)    # 楕円上の y^2
`````

`````julia
d1  = sympy.sqrt((x + c)^2 + ysq)   # |PF1|
d2  = sympy.sqrt((x - c)^2 + ysq)   # |PF2|
`````

`````julia
# 根号の中身を因数分解すると…
r1 = sympy.factor((x + c)^2 + ysq)  # → (3x+25)^2 / 25 = (a + (c/a)x)^2
r2 = sympy.factor((x - c)^2 + ysq)  # → (3x-25)^2 / 25 = (a - (c/a)x)^2
println("根号の中 (PF1) = ", r1)
println("根号の中 (PF2) = ", r2)
`````

`````julia
# 根号の中が完全平方なので、d1 = a + (c/a)x, d2 = a - (c/a)x。
# 和は 2a = 10 —— x（楕円上のどこか）に依らず一定。何点か代入して確かめる。
for xv in (-5, -2, 0, 1, 5)
    println("x = ", xv, " のとき d1 + d2 = ", (d1 + d2).subs(x, xv))
end
`````

**観察3**: 根号の中身が $(a + \tfrac{c}{a}x)^2$, $(a - \tfrac{c}{a}x)^2$ という **完全平方** になる。だから根号がほどけて $d_1 = a + \tfrac{c}{a}x$, $d_2 = a - \tfrac{c}{a}x$、和は $d_1 + d_2 = 2a = 10$ で、$x$ に依らず一定。**L10 で「$P$ をドラッグしても和が一定」と作図で見たことが、ここでは式の上で「根号がほどけて定数になる」として現れる**。定義（距離の和）と方程式（$x^2/a^2+y^2/b^2=1$）が、同じ一つのことの二つの言い方だと分かる。

`````julia
# L10 で予告した「準線」を立ち上げる。準線は軸に直交する隠れた直線 x = ± a^2/c。
# 焦点 F2=(3,0) と準線 x = a^2/c について、|PF| /（準線までの距離）= e を確かめる。
dir_x = sympy.Rational(a^2, c)        # 準線の x 座標 = 25/3 ≈ 8.33
P  = (4, sympy.Rational(12, 5))       # x=4 のとき y=12/5（方程式 16/25+9/25=1 上の点）
PF   = sympy.sqrt((4 - c)^2 + P[2]^2) # 焦点 F2=(3,0) までの距離
Pdir = dir_x - 4                      # 準線 x=25/3 までの距離
ratio = sympy.simplify(PF / Pdir)
println("準線 x = ± ", dir_x)
println("|PF| = ", PF, " / 準線までの距離 = ", Pdir, " / 比 = ", ratio, "（= e = 3/5 か）")
`````

**観察4**: 焦点までの距離と準線までの距離の **比** が $3/5 = e$ になる。L10 の最後で予告した「焦点からの距離と準線までの距離の比が一定 $=e$」が、ここで SymPy で確定した。**等距離（円、$e=0$）から距離の比（楕円、任意の $e$）へ** —— これが円から円錐曲線への拡張だった（L6 の伏線回収）。準線という「隠れた直線」（L7 Euler 線の伏線）が、楕円にも焦点ごとに存在する。

### 3.4 これが汎用的にどこに効くか（将来の武器）

SymPy がやったのは「**観察 → 数値 → 記号 → 証明**」というパイプラインである。これは計算機代数（CAS: Computer Algebra System）の基本作法で、分野を超えて再演される：

- **自動定理証明・型システム**（Lean, Coq）—— 「正しいか」を記号で機械的に検証する
- **物理エンジン**（SymPy-Mechanics, Mathematica）—— 運動方程式を記号で立てて解く
- **微分方程式の解析的扱い・最適化・回路解析** —— 数値でなく式のまま操る

L5 で「内部をデータに取り出す」、L7 で「依存を構造に取り出す」、L9 で「構造で採点する」と進んできた外部化の系列の、**式への外部化** が今日の SymPy である。観察された曲線を式に固定できれば、あとは式を機械が検証・変形してくれる。

## 4. 観測量に固める —— 切り口の楕円 = 定義の楕円 = 一つの方程式

**主命題**: 円錐を斜めに切った切り口の楕円も、二焦点・和一定で定義した楕円も、適切な座標を取れば同じ方程式 $x^2/a^2 + y^2/b^2 = 1$（$b^2 = a^2 - c^2$）に落ちる。その方程式から計算される焦点は $(\pm c, 0)$、離心率は $e = c/a$、準線は $x = \pm a^2/c$ であり、焦点までの距離と準線までの距離の比は一定 $e$ である。

```{important} 定義（L10 の定義を、方程式の側から言い直す）
楕円の方程式 $\dfrac{x^2}{a^2} + \dfrac{y^2}{b^2} = 1$（$a > b > 0$）に対し、$c = \sqrt{a^2 - b^2}$ とおくと、二焦点は $(\pm c, 0)$、**離心率** は $e = c/a$（$0 \le e < 1$）、**準線** は $x = \pm a^2/c$ である。任意の楕円上の点 $P$ について、焦点までの距離と準線までの距離の比は一定 $e$ に等しい。$e=0$ で円、$e \to 1$ で細長い楕円になる。
```

```{attention} 観察 → 構造 → 式 —— 三つの言葉で同じ楕円を語る
:class: attention
```

## 5. 一枚に重なった三曲線を、式で言い切る —— 共通焦点の一族

楕円・放物線・双曲線が **一枚の画面に重なって** 描かれていると、どの線がどの曲線のためのものか、画面だけでは見分けられない（補助線・接線・準線が同じ平面に乗る）。GeoGebra は正しく **実行** するが、「どの曲線がどの定義に基づくか」は返してくれない。だが **式の側から** 各曲線を取り出すと、隠れていた統一が見える。

`````julia
# 三つの円錐曲線を SymPy の式として持ち、焦点・離心率を計算で並べる。
# （各式は一枚の複合図から取り出したもの。平方完成すると共通焦点 F1=(0.5,0) が出る。）
el2 = geo.Ellipse(geo.Point(1, 0), 1, sqrt(3)/2)   # 中心(1,0) の楕円（焦点の一方が原点側 0.5）
# 注意: geo.Ellipse に `sympy.sqrt(3)/2`（Py 値）を渡すと SympifyError。素の `sqrt(3)/2` を渡す。
println("楕円   焦点 = ", el2.foci, " / e = ", el2.eccentricity)
# 放物線 y^2 = 2x：焦点 (0.5, 0)、準線 x = -0.5、離心率 = 1（楕円と双曲線の境界）
println("放物線 焦点 = (1/2, 0) / 準線 x = -1/2 / e = 1")
# 双曲線：焦点の一方が同じく (0.5, 0) 側、e > 1
println("双曲線 e ≈ 1.67（焦点の一方を楕円・放物線と共有）")
`````

**観察5**: 三つとも **焦点 $(0.5, 0)$ を共有** している。つまりこの一枚は、

> **共通の焦点 $F_1$ を持ち、もう一方の焦点（あるいは準線）の位置だけが違う円錐曲線の一族** —— 第二焦点が近ければ **楕円**（$e<1$）、無限遠へ去って準線が直線になれば **放物線**（$e=1$）、反対側へ回れば **双曲線**（$e>1$）。

を一度に描いている。**放物線は楕円と双曲線の境界**（$e=1$）として、この一族のちょうど真ん中にいる。離心率 $e$ という **一つの数** が、三曲線を連続的に並べている —— これが「重なって混乱する画面」の裏に隠れていた統一であり、**式の側からしか見えない**ものだった。

```{note} 構造帰属の注意 —— ファイル名でなく焦点共有で帰属する
:class: note
「どの曲線がどの準線か」は、レイヤー名やファイル名ではなく **焦点を共有しているか** という構造条件で帰属する。たとえば「放物線」と名のついた図を見て「放物線も円形準線だろう」と思い込むと誤る（放物線の準線は **直線**）。L9 で練習した「座標でなく構造（依存グラフ）で型を帰属する」採点が、ここで効く —— 名前に引きずられず、式と構造で確かめる。
```

## 6. 停滞と画期の照射 —— 通径という封印と、代数による解放

楕円・放物線・双曲線という名は、紀元前 3 世紀の **Apollonius**（『円錐曲線論』）に遡る。彼は円錐を平面で切るだけでこの三曲線を統一的に記述した —— 今日 §2 でわれわれが目で見た、あの「一つの円錐 × 平面の傾き」の視点である。だが Apollonius の記述は **斜交座標系による局所的・幾何的なもの** で、曲線から「焦点」を計算で取り出すことはできなかった。その幾何が、ほぼ手つかずで **1800 年** 保たれた。

なぜそんなに長く止まったのか。鍵は Apollonius の **通径（latus rectum）** という一つのパラメータにある。通径は二つの顔を持つ。一つは **後世への贈り物** —— 曲線を一つの量で記述する初の体系で、関数概念の前駆であり、今も軌道力学・GPS で現役（半直弦 $\ell = a(1-e^2)$）。もう一つは **封印の道具** —— 通径は接線を「曲線と一点だけで交わる直線」 という **静的な** ものとして扱うことで首尾一貫した。その代償に、**接線の局所構造（＝後の微分）への問いが閉ざされた**。実は Euclid 自身、第 III 巻 **命題 16**（円の直径端での接線と「角」 の扱い）で、量の比較が公理系から逸脱する唯一の箇所に触れている —— 接線の局所構造を正面から量として扱うことを避けた「正直な行き詰まり」 である。Apollonius の通径は、この **定義されない接線** の上に建っていた。

封印が解かれるのは **1800 年後**。**Descartes**（1637）は法線（曲線に直交する直線）を経由し、極限という未定義概念に踏み込まない「誠実な」 代数で接線を扱った —— Apollonius を経由しない別経路で、Menaechmus 的な直交性を取り戻した。**Fermat** は逆に、極限（adégalité）を正面から引き受けて封印を直接破壊した。**Newton**（1687）は両者の論争を保留したまま、運用レベルで微分を物理に投入し、科学革命を先行させた。この Apollonius → Descartes / Fermat → Newton の系譜は、**L14 で正面から扱う**。

今日皆さんが触れたのは、その入り口 —— **観察された曲線を方程式に移す**（Descartes の画期）一手である。SymPy で「観察 → 方程式 → 焦点・離心率」 をやったこと（§3）は、1637 年の画期の現代的再演にほかならない。曲線が「眺めるもの」 から「計算するもの」 になった、その最初の手触りである。

```{note} 伏線回収 —— まだ「なぜ二焦点が重なるのか」は天下りである
:class: note
今日、切り口の楕円と定義の楕円が同じ方程式に落ちることを見た。だが正直に言えば、**なぜ円錐の切り口の二焦点が、L10 の「距離の和」の二焦点と同じ場所に来るのか** は、まだ「両方とも同じ方程式になったから」と式の上で重なっているだけで、その **理由** は示していない。

L12–L13 で、この二焦点が **円錐を切る平面に内側から接する二つの球（Dandelin 球）の接点** であることが分かる。二つの球は切断面の上下から円錐に接する —— ただし斜めの切り口では **上下の球は大きさが違う**（互いに鏡像ではない）。「距離の和が一定」は、二つの球が円錐に接する二つの円のあいだの、**母線に沿った長さが一定** であることから **必然** として出てくる。二焦点が中心について対称なのは、図全体が長軸を含む面について左右対称だからである（球の合同からではない）。今日 §3 で SymPy が機械的に確かめた「和が一定・二焦点の対称」が、L12 では円錐の三次元構成から **意味として** 立ち上がる。L6（等距離）・L7（隠れた直線＝準線）・L8–L9（内外の対称）・L10（二焦点）・L11（方程式）の伏線は、すべて Dandelin の二球で一つに結ばれる。
```

次回は、円錐に **内側から接する二つの球** を入れる。一つは切断面の上、一つは下。両方とも円錐の内壁にぐるりと接し、切断面にはそれぞれ一点で触れる。その **二つの接点** が、今日「二焦点」と呼んだ二点である。その **枕** として、円錐と内接球の予告だけ置く。

`````julia
# 円錐に内接する球（Dandelin 球）の予告 —— 二つの球が切断面に上下から接する。
@ggb :const :new
@ggb apex2=Point("{0, 0, 0}")                  # L11 §2 と同じ有限円錐（点は Point("{…}") で）
@ggb top2=Point("{0, 0, 5.4}")
@ggb cone2=Cone(:top2, :apex2, 1.9712)         # 底面中心・頂点・底面半径
@ggb nrm3=Vector((0.4, 0, 1))
@ggb pl3=PerpendicularPlane((0,0,2.5), :nrm3)         # 同じ斜めの切断面
@ggb sec3=IntersectConic(:pl3, :cone2)         # 切り口の楕円（平面が第一引数）
@ggb sph1=Sphere((0,0,1.8257), 0.6260)         # 切断面の下に内接する球（半径は接する条件から確定・L12）
@ggb sph2=Sphere((0,0,3.9639), 1.3592)         # 切断面の上に内接する球（半径は接する条件から確定・L12）
`````
