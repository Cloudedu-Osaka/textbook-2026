## 1. 束にする —— 一本では見えなかったものが浮かび上がる

### 1.1 予測 —— 掃いた跡には何が残るか

L14 の枕で、円 `lk`（中心 $F_2$・半径 4）の上の点 $Q$ と、円の中の点 $F_1 = (1.5, 0)$ の垂直二等分線 `m` を一本だけ引いた。$Q$ をドラッグすると `m` が動く —— その **掃いていく跡** に何が浮かぶか、予測は書き留めてあるだろうか。まだなら、いま書く。

`````julia
using Pkg
Pkg.activate("../..")
Pkg.resolve()
Pkg.instantiate()
using GeoGebra
ENV["GGB_DIRECT_TRANSPORT"] = "true"
`````

`````julia
inject_applet()
`````

`````julia
# L14 の枕の再現 + 今日の主役たち。規律1: 各オブジェクトを一意のシンボルに束縛。
@ggb :const :new
@ggb F_2=(0, 0)
@ggb lk=Circle(:F_2, 4)                    # 円（半径 4 = のちに 2a になる）
@ggb F_1=(1.5, 0)                          # 円の中の点（あとでドラッグする）
@ggb Q=Point(:lk)                          # 円の上の点
@ggb m=PerpendicularBisector(:F_1, :Q)     # 垂直二等分線（これが「一本」）
@ggb rq=Segment(:F_2, :Q)                  # 中心から Q への半径
@ggb P=Intersect(:m, :rq)                  # 半径と垂直二等分線の交点（今日の主役）
`````

`````julia
# P の軌跡（Q を一周させたときの P の跡）と、m の束の跡を見る。
# アプレット上で m を右クリック →「残像を表示」(Trace On) にしてから Q をドラッグすると
# 「束」が掃く跡が見える。軌跡コマンドでも確かめる:
@ggb loc=Locus(:P, :Q)
`````

**観察1**: $P$ の軌跡 `loc` は **楕円**。そして残像にした `m` の束は、その楕円を **外側から包む** —— どの一本も楕円に **接して** いる（L14 の言葉で言えば、束のどの一本も接線であり、楕円は束が浮かび上がらせた **包絡線** である）。一本の直線はただの直線なのに、束が曲線を運んでいた。

### 1.2 なぜ楕円か —— 一行で言い切る

$P$ は垂直二等分線 `m` の上にいるから $|PF_1| = |PQ|$。$P$ は半径 $F_2Q$ の上にもいるから

$$|PF_1| + |PF_2| = |PQ| + |PF_2| = |F_2Q| = 4 \quad (\text{半径・一定}).$$

**和が一定 —— L10 の楕円の定義そのもの**である。$2a = 4$、焦点は $F_1$ と $F_2$。L10 では紐で楕円を描いた。今日は **紐の長さが円の半径に化けている** —— 紐の一端を円周まで伸ばしておいて、垂直二等分線で「折り返した」だけである。

`````julia
# 一行証明を数値でも: P をどこへ動かしても和は半径 4 のまま。
@ggb d1=Distance(:P, :F_1)
@ggb d2=Distance(:P, :F_2)
@ggb s="d1 + d2"                             # = 4（Q をドラッグしても一定）
`````

**観察2**: `s = 4` から動かない。ついでに `m` と `loc` の交点を GeoGebra に数えさせてみると（L14 §2 の再演）、undefined の席が混じる —— `m` は `loc` に **接して** いる（重解）。L14 で言葉にした「接する」が、束の一本一本で起こっている。

**観察3**: (1) $F_1 = F_2$ のとき軌跡は **円**（$e = 0$）。(2) 円周に近づくほど楕円は細長くなる（$e \to 1$）。(3) $F_1$ を円の外に出すと、軌跡は **双曲線** になる —— このとき $P$ は $|PF_2| - |PF_1| = \pm 4$（**差** が一定・半径のまま）を満たす（垂直二等分線の同じ一行が、和ではなく差を作る。どちら側の枝かは $Q$ の位置による）。円 → 楕円 → 双曲線の全族が、**点を一つ動かすだけ** で連続に生まれた。L13 では切断面の傾きが、今日は $F_1$ の位置が、一族を貫くつまみである。

## 3. 1964 年 —— 古い道具しか使わずに、新しい証明が生まれた

### 3.1 失われた講義

1964 年 3 月、Caltech の一年生向け講義で Feynman は「惑星の太陽のまわりの運動」と題した特別講義をした。目標は Newton の主定理 —— **逆二乗の引力から、軌道が楕円になることを導く** —— を、**微積分を使わずに** 示すこと。Feynman は Newton の『Principia』（1687）の幾何学的証明を途中まで追い、Newton が円錐曲線の古典的性質（まさに Apollonius 流）に頼る箇所で「私はこの性質をよく知らないので、自分の証明を作った」と道を逸れる。そこで彼が組み立てたのが、**速度の図** を使う初等幾何の証明である。Feynman 自身の言葉 —— 「学生にこれをやってみせるのはひどく難しい。だが私は最も簡素なものを見つけようとした」。

骨子は五段でスケッチできる（完全な再構成は参考文献の Goodstein 夫妻の本に）:

1. 軌道を **等しい角度** で分割する（Newton は等しい **時間** で分割した。**等角度に切り替えるのが Feynman の本質的な一手**）。
2. 逆二乗則と等角度分割を組み合わせると、一区間ごとの速度の変化 $\Delta v$ が、**どの区間でも同じ大きさ** になる。
3. だから速度の矢印を一点から並べて描くと（この図を **ホドグラフ** と呼ぶ）、矢印の先端は等間隔の折れ線 —— 極限で **完全な円** —— を描く。円の中心が原点からずれている分が、軌道の離心率になる。
4. この速度の円の図を 90° 回転して置き直すと、そこに現れるのが —— **円と、円の中の一点と、垂直二等分線の束**。今日 §1 でやった、あの構成である。
5. その構成から軌道を再構成すると、楕円が得られる（§1.2 の一行証明がそのまま働く）。

ここで L10 の Kepler を思い出してほしい。Kepler は **速度の変化のパターン** を観測データの手計算から抽出し（面積の法則・1602）、楕円という **形** はその三年後（1605）に、運動と辻褄の合う形として後から見つけた —— 形より運動が先だった。Feynman のホドグラフは、**Kepler が数の表の中に見た速度のパターンを、定規とコンパスで作図できる一つの円** に変えたものである。三百年を隔てて、同じ「速度が先、形が後」の道筋が、計算から作図へ翻訳された。

この講義は録音とメモが長く行方不明で、「失われた講義」と呼ばれた。1990 年代に発掘・再構成されて出版されている。

### 3.2 二千年の見落とし・二例目 —— ただし、帰属は正確に

L12 §6 で Dandelin（1822）についてこう言った —— 円錐・平面・球というユークリッド幾何の範囲だけでできていて、古代ギリシア人が見つけていてもおかしくなかったのに、見つかるまで二千百年かかった、と。Feynman の証明も同じ型である。使った道具は円と直線と相似 —— **Euclid が全部持っていた**。Newton 自身が『Principia』の中でやっていてもおかしくなかった。それでも、この形の証明は 20 世紀後半まで現れなかった。

ここで一つ、**言いすぎない** 練習をする（L14 §6 の作法の実演である）。「Feynman が円形準線を発見した」と書きたくなる —— が、それは **帰属として言いすぎ** である。速度図が円になることは 19 世紀に Hamilton が見つけていた。円形準線（Leitkreis）も 19 世紀の総合幾何では標準装備だった。Feynman の功績は「発見」ではなく、ばらばらに存在していた部品を **微積分ぬきの一本の証明に編み直し、作図できる装置として復権させた** ことである —— それで十分に偉大であり、正確に言った方が、彼が何をしたのかがむしろ鮮明になる（Kepler の功績が「点の発見」ではなく「命名と同定」だったのと同じ構図である —— L13 §5.1）。

- **Dandelin 1822**: 道具は Euclid のまま。見つかったのは Apollonius から二千百年後。
- **Feynman 1964**: 部品は Euclid と 19 世紀のまま。編み直されたのは Newton から三百年近く後。

**画期は、最新の道具を要求しない**。要求するのは、見る場所を変えること —— Dandelin は円錐の「中の球」を、Feynman は「速度の空間」を見た。逆に言えば、**道具が全部そろっていても、見る場所が固定されていれば、二千年でも見落とし続ける**。これが、皆さんに渡す最後の観察である。

### 4.1 議論の共通レンズ —— 三角構造

発表を聴くとき・議論するときの共通レンズとして、学期を貫いた **三角構造** を確認しておく:

```
   譲らない構成                     実を取る譲歩
   （厳密さを手放さない側）          （厳密さを一部保留して前進する側）
        \                                  /
         \                                /
          \_________ 運用統合 _________/
          （両方の良いところを回して前進させる誰か）
```

- Euclid は「入り込めない」で譲らず（L14）、Apollonius は 487 命題で体系を譲らず（L11–L13）—— そして体系の外（準線・重解）が二千年見えなくなった。
- Descartes は代数で譲らない基盤を作り、Fermat は「重なる二交点」という危うい譲歩（adequality）で実を取り、Newton がそれらを **決着を保留したまま運用統合** して力学が走り出した（L14 §4.3）。
- Dandelin と Feynman は、固定された「見る場所」の外に立った（L12・今日）。

**停滞は、多くの場合、誰かが悪いのではなく、この三角のどこかが欠けている状態** である —— 譲らない側しかいない（動けない）、譲歩だけがある（土台が腐る）、統合者が現れない（両者が並走したまま）。皆さんが転生した先の人物も、必ずこの三角のどこかに立っていたはずである。

## 5. 結び —— 半円の直角から、ここまで

最終回なので、種明かしを一つ。今日の構成の芯 —— 垂直二等分線と、半径との交点 —— を支えていたのは、$|PF_1| = |PQ|$ という **等距離** だった。等距離を運ぶ垂直二等分線は L4 で外心を組み立てた道具であり、Thales の「半円に内接する角は直角」は L14 の接線作図で最後まで働いていた。**第 4 回に置いた最小の公準が、第 15 回の Feynman まで、一度も途切れずに通っている**。

そして Dandelin と Feynman が示した通り、道具がそろっていることと、画期が起こることのあいだには、二千年の隙間が空きうる。隙間を埋めるのは新しい道具ではなく、見る場所の変更だった。円錐曲線という二千三百年前に始まった分野で、19 世紀と 20 世紀に初等幾何の新発見があった —— 二千年をかけても発見できない美しい真実が、まだ残っている（L12 §6）。皆さんが今日語った転生 —— 過去の一人の隣に立ち、その人の半歩先を一緒に見ようとした練習 —— は、その「見る場所」を動かす稽古である。

この講義で渡したかったのは、円錐曲線でも Julia でもなく、この診断装置である。どこかで停滞に出会ったら —— それは必ず出会う —— 三角を描き、反証条件を書き、冷間で一撃を探し、そして見る場所を疑ってほしい。Menaechmus の直交から Feynman の速度の円まで、この物語の円環は閉じていない。今も開いている —— 次の一筆を書き加えるのが誰かは、決まっていない。
